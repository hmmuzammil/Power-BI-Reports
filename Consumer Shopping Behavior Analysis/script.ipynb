```python
import pandas as pd
```

```python
df= pd.read_csv(r"D:\Downloads\customer_shopping_behavior.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns= df.columns.str.lower()
df.columns= df.columns.str.replace(' ', '_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
#Creaate column for Age Group
labels = ['Young', 'Adult', 'Middle-Aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age', 'age_group']].head()
```

```python
#Create column for frequency_purchase_days
frequency_mapping = {
    'Fornightly': 14,
    'Monthly': 30,
    'Weekly': 7,
    'Quarterly': 90,
    'Anually': 365,
    'Bi-Weekly': 14
}

df['frequency_puchase_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_puchase_days','frequency_of_purchases']].head()
```

```python
df.drop('promo_code_used', axis=1, inplace=True)
```
